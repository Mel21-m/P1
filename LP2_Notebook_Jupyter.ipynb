## 3. Importación de datos

A diferencia del ejemplo teórico con datos creados dentro del notebook, aquí trabajaremos con un **archivo externo**.

## 4. Control básico de calidad

Un reporte reproducible debe dejar evidencia de las verificaciones realizadas sobre los datos.

Buscaremos:

- valores faltantes;
- observaciones duplicadas;
- rangos potencialmente extraños.

`````python
# COMPLETE: cantidad de valores faltantes por variable
faltantes = datos.isna()
faltantes
`````

`````python
# COMPLETE: número de filas duplicadas
n_duplicados = datos.duplicated().sum()
print("Filas duplicadas:", n_duplicados)
`````

### Actividad 4 — Limpieza

Genere un objeto llamado `datos_limpios` que:

1. sea una copia de `datos`;
2. elimine filas duplicadas;
3. complete los valores faltantes de `temperatura_c` y `humedad_suelo_pct`
   usando la **mediana del módulo correspondiente**.

Al final, demuestre que ya no quedan faltantes ni duplicados.

`````python
# COMPLETE
datos_limpios = datos.copy()

datos_limpios = datos_limpios.drop_duplicates()

for variable in ["temperatura_c", "humedad_suelo_pct"]:
    datos_limpios[variable] = (
        datos_limpios.groupby("modulo")[variable]
        .transform(lambda x: x.fillna(x.median()))
    )

print("Faltantes totales:", datos_limpios.isna().sum().sum())
print("Duplicados:", datos_limpios.duplicated().sum())
`````

## 5. Construcción de una variable derivada

Definiremos como condición ambiental para **revisión** aquella observación que cumpla al menos una de estas condiciones:

- temperatura menor de 18 °C o mayor de 27 °C;
- humedad del suelo menor de 38 %.

Cree una variable `estado` con las categorías **"Adecuado"** y **"Revisar"**.

`````python
# Solución
condicion_revision = (
    (datos_limpios["temperatura_c"] < 18) |
    (datos_limpios["temperatura_c"] > 27) |
    (datos_limpios["humedad_suelo_pct"] < 38)
)

datos_limpios["estado"] = np.where(
    condicion_revision,
    "Revisar",
    "Adecuado"
)

datos_limpios[["fecha", "modulo", "temperatura_c", "humedad_suelo_pct", "estado"]].head(20)
`````

## 6. Indicadores principales (KPIs)

Calcule:

- número de observaciones válidas;
- temperatura promedio;
- consumo total de riego;
- porcentaje de observaciones clasificadas como `"Revisar"`.

`````python
# SOLUCIÓN
n = len(datos_limpios)
temp_prom = datos_limpios["temperatura_c"].mean()
riego_total = datos_limpios["riego_litros"].sum()
pct_revision = datos_limpios["estado"].eq("Revisar").mean() * 100

kpis = pd.DataFrame({
    "Indicador": [
        "Observaciones",
        "Temperatura promedio (°C)",
        "Riego total (L)",
        "Observaciones para revisión (%)"
    ],
    "Resultado": [
        n,
        round(temp_prom, 2),
        round(riego_total, 1),
        round(pct_revision, 1)
    ]
})

kpis
`````

## 7. Resumen por módulo

Construya una tabla con una fila por módulo y las siguientes columnas:

- `temperatura_media`;
- `humedad_suelo_media`;
- `riego_medio`;
- `altura_media`;
- `n_observaciones`.

Ordene el resultado desde el mayor al menor `riego_medio`.

`````python
# COMPLETE
resumen_modulo = (
    datos_limpios
    .groupby("modulo")
    .agg(
        temperatura_media=("temperatura_c", "mean"),
        humedad_suelo_media=("humedad_suelo_pct", "mean"),
        riego_medio=("riego_litros", "mean"),
        altura_media=("altura_cm", "mean"),
        n_observaciones=("modulo", "size")
    )
    .sort_values("riego_medio",ascending=False)
)

resumen_modulo.round(2)
`````

## 8. Relación entre variables

Calcule la matriz de correlaciones de las variables numéricas seleccionadas:

`temperatura_c`, `humedad_ambiente_pct`, `humedad_suelo_pct`, `riego_litros`, `altura_cm`.

Luego identifique la correlación entre **temperatura y riego**.

`````python
# SOLUCIÓN
variables = [
    "temperatura_c",
    "humedad_ambiente_pct",
    "humedad_suelo_pct",
    "riego_litros",
    "altura_cm"
]

correlaciones = datos_limpios[variables].corr()   #en r tambien es corr
correlaciones.round(2)
`````

`````python
# SOLUCIÓN
r_temp_riego = correlaciones.loc["temperatura_c", "riego_litros"]
print(f"Correlación temperatura-riego: {r_temp_riego:.3f}")
`````

## 9. Visualización 1 — Evolución de la temperatura

Construya un gráfico de líneas con:

- fecha en el eje X;
- temperatura en el eje Y;
- una línea por módulo;
- título, ejes, leyenda y cuadrícula;
- tamaño adecuado para un reporte.

`````python
# SOLUCIÓN
fig, ax = plt.subplots(figsize=(11, 5))

for modulo, grupo in datos_limpios.groupby("modulo"):
    ax.plot(
        grupo["fecha"],
        grupo["temperatura_c"],
        marker="o",
        markersize=3,
        linewidth=1.6,
        label=modulo
    )

ax.set_title("Evolución diaria de la temperatura por módulo")
ax.set_xlabel("Fecha")
ax.set_ylabel("Temperatura (°C)")
ax.legend(title="Módulo")
ax.grid(axis="y", alpha=0.25)
ax.grid(axis="x", alpha=0.25)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
`````

## 10. Visualización 2 — Comparación del riego medio

Construya un gráfico de barras del riego promedio por módulo.

`````python
# SOLUCIÓN
riego_modulo = (
    datos_limpios.groupby("modulo")["riego_litros"]
    .mean()
    .sort_values(ascending=False)
)

ax = riego_modulo.plot(
    kind="bar",
    figsize=(8, 4),
    color= ["#70362A","#A85F4F","#CF7F6D" ]
)

ax.set_title("Riego promedio por módulo")
ax.set_xlabel("Módulo")
ax.set_ylabel("Riego promedio (L)")
ax.grid(axis="y", alpha=0.25)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
`````

## 11. Visualización 3 — Temperatura vs. riego

Construya un diagrama de dispersión que permita estudiar si el riego aumenta cuando aumenta la temperatura.  
Use una marca diferente para cada módulo o grafique cada módulo por separado dentro del mismo eje.

`````python
# COMPLETE

for modulo, grupo in datos_limpios.groupby("modulo"):
    ax.scatter(
        grupo["temperatura_c"],
        grupo["riego_litros"],
        alpha=0.75,
        label=modulo
    )

ax.set_title("Relación entre temperatura y riego")
ax.set_xlabel("Temperatura (°C)")
ax.set_ylabel("Riego (L)")
ax.legend(title="Módulo")
ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()
`````

4. El análisis por módulo muestra diferencias sistemáticas entre los cultivos, por lo que no conviene interpretar el consumo de agua sin considerar el módulo/cultivo.

### Recomendación
Mantener un seguimiento conjunto de temperatura, humedad del suelo y riego. Las alertas no deberían basarse en una sola variable: una regla reproducible dentro del notebook permite identificar automáticamente observaciones que requieren revisión.
